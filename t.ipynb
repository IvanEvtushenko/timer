```python
def hello(st):
    print(st)
    return 0

c = 0

def cnt(f): 
    def wrap(*args, **kwargs):
        global c
        c += 1
        return f(*args, **kwargs), c

    return wrap

hello = cnt(hello)

print(hello("6"))
print(c)
print(hello("9"))
print(c)

```

```python
def fib(n):
    n_1, n_2 = 1, 1
    for i in range(n):
        yield n_1
        n_1, n_2 = n_2, n_1 + n_2


print(", ".join((str(x) for x in fib(10))))
```
